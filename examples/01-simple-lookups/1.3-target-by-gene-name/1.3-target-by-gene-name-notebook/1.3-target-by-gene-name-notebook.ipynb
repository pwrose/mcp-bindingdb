````python
# Call 2 · get_target · ran 2026-11-24 04:54:59 UTC
await replay.call(2, 'get_target', {'target_id': 520},
                  sha256='575528fd2ce471ce69831bf7994714a77f700056c62fe000520a8635bfb43604')
````

````python
# Call 3 · run_sql · ran 2026-11-24 04:54:59 UTC
sql = """
    -- All BindingDB targets on UniProt P00533 (human EGFR): wild type, mutants and constructs
    SELECT polymerid AS target_id, uniprot_raw, count(*) AS n_measurements, count(DISTINCT monomerid) AS n_compounds
    FROM activity
    WHERE uniprot = 'P00533'
    GROUP BY ALL
    ORDER BY n_measurements DESC, target_id
    
"""
await replay.call(3, 'run_sql', {'sql': sql},
                  sha256='aec91062a91dbab22645fcbe178dfc48416e977f52e38ec9e898a4546db3c3b4')
````

````python
# Call 4 · run_sql · ran 2026-11-24 04:54:59 UTC
sql = """
    -- Whole-gene total for P00533, and EGFR targets in other organisms
    SELECT 'P00533, all variants' AS target, 'Homo sapiens' AS organism, count(DISTINCT polymerid) AS n_targets, count(*) AS n_measurements, count(DISTINCT monomerid) AS n_compounds
    FROM activity WHERE uniprot = 'P00533'
    UNION ALL
    SELECT uniprot_raw || ' ' || target_name, organism, count(DISTINCT polymerid), count(*), count(DISTINCT monomerid)
    FROM activity
    WHERE organism <> 'Homo sapiens' AND target_name ILIKE 'Epidermal growth factor receptor%'
    GROUP BY ALL
    ORDER BY n_measurements DESC
    
"""
await replay.call(4, 'run_sql', {'sql': sql},
                  sha256='63c4c798b52b52c3bc12335c747e9539b2043faacd83fecef3f7f4c7ea26b2ce')
````

**🧠 Assistant**

Human EGFR (epidermal growth factor receptor, UniProt **P00533**) is BindingDB target **520**, with **23,267 measurements on 14,127 compounds**. Most of these are IC50 values:

| Affinity type | Measurements | Compounds | Best pAffinity |
|---|---|---|---|
| IC50 | 20,998 | 13,814 | 11.7 |
| Kd | 1,593 | 180 | 10.0 |
| Ki | 409 | 288 | 11.22 |
| EC50 | 267 | 147 | 9.0 |

Other facts from the target record: 1,210 residues, ChEMBL id CHEMBL1955, 278 linked PDB structures (for example 1M17, erlotinib bound to the kinase domain), and synonyms including ERBB1, HER1 and EGFR_HUMAN.

**One gene, many BindingDB targets.** Target 520 is only the wild-type record. BindingDB stores every **mutant and construct** as a separate target. Searching by the accession P00533 returns **58 target records**, and the 56 of them that have data hold **31,621 measurements on 16,406 compounds** in total. The largest are:

| Target id | Variant | Measurements | Compounds |
|---|---|---|---|
| 520 | wild type | 23,267 | 14,127 |
| 60 | T790M / L858R | 1,536 | 1,276 |
| 1058 | L858R / T790M / C797S | 703 | 679 |
| 1214 | L858R / C797S | 684 | 683 |
| 51 | L858R | 678 | 646 |
| 1655 | exon-19 deletion (Δ746-750) / C797S | 665 | 664 |
| 1198 | T790M / C797S (residue 19 deleted) | 478 | 451 |
| 1197 | C797S (residue 19 deleted) | 428 | 415 |
| 1253 | exon-20 insertion NPG after 770 | 415 | 410 |
| 1252 | exon-20 insertion SVD after 770 | 302 | 282 |
| 535 | exon-19 deletion (Δ746-750) | 246 | 245 |
| 243 | T790M | 229 | 213 |
| 1150 | kinase domain 695-1022 | 142 | 124 |

The variant is written in the target name and in `uniprot_raw`: `P00533[T790M,L858R]` is a double mutant, `P00533[1-745,751-1210]` is the exon-19 deletion (residues 746-750 missing), `P00533[1-770,'NPG',771-1210]` is an exon-20 insertion, and `P00533[695-1022]` is a construct covering residues 695-1022. The clinically important resistance mutations (T790M, C797S), the activating mutations (L858R, exon-19 deletions) and the exon-20 insertions each have their own data. Note that the same triple mutant can appear more than once: L858R/T790M/C797S is target 1058 (703 measurements) and also target 1788 (`P00533[T790M,C797S,L858R]`, 275 measurements).

**Two cautions about the organism filter.**

- Without an organism filter, a name search for EGFR also returns **mouse Egfr** (Q01279, 86 measurements). Add "human" (`organism='Homo sapiens'`) when you mean the human protein.
- But the filter also has a cost: **8 of the 58 P00533 records have no organism recorded**, and they drop out of a search filtered on Homo sapiens. They hold 1,314 measurements, including the exon-19 deletion/C797S mutant (target 1655, 665 measurements) and the second L858R/T790M/C797S record (target 1788, 275). Searching by the accession P00533 finds them all.

**The lesson.** Decide which target you mean before you ask for ligands. Use the **wild-type id (520)** for the classic "EGFR inhibitor" question. Use a **mutant id** for resistance questions, and check for duplicate records of the same mutant. Use the **UniProt accession with variants included** when you want everything measured on the gene.

**How this was answered**

- `search_targets` with query "EGFR" and `organism='Homo sapiens'`: matches gene names, protein names, UniProt entry names and accessions. Exact matches and the targets with the most data rank first.
- `get_target` with id 520: identifiers, synonyms, PDB ids and counts by affinity type.
- `run_sql`: counted all targets that share UniProt P00533 and their measurements, and the EGFR targets recorded under another organism (or none):

```sql
SELECT polymerid AS target_id, uniprot_raw, count(*) AS n_measurements, count(DISTINCT monomerid) AS n_compounds
FROM activity
WHERE uniprot = 'P00533'
GROUP BY ALL
ORDER BY n_measurements DESC, target_id
```

- `search_targets` with query "P00533": every target record on the accession, with its organism field.
