## Exchange 1

**👤 User**

What does BindingDB know about imatinib?

````python
# Call 1 · search_compounds · ran 2026-11-24 04:54:59 UTC
await replay.call(1, 'search_compounds', {'query': 'imatinib'},
                  sha256='1f2a50bf819de48009ec8144d0b21eef4aca07d81ee80cd8718d9c5f2d42211d')
````

````python
# Call 2 · get_compound · ran 2026-11-24 04:55:00 UTC
await replay.call(2, 'get_compound', {'monomerid': 13530},
                  sha256='1fb7788c17f88a1c747d2878b057fe40de8b095e2bb102bb6f3c6863fc12f468')
````

````python
# Call 3 · run_sql · ran 2026-11-24 04:55:00 UTC
sql = """
    -- Imatinib (BDBM13530): median vs best pAffinity per target, for its clinical targets and the top 'best value' targets
    SELECT target_name, uniprot_raw, count(*) AS n,
           sum(CASE WHEN relation = '>' THEN 1 ELSE 0 END) AS n_censored,
           round(median(p_affinity), 2) AS median_p,
           round(max(p_affinity), 2) AS best_p
    FROM activity
    WHERE monomerid = 13530
      AND uniprot_raw IN ('P00519', 'P10721', 'P16234', 'P04626', 'P00533', 'Q08345')
      AND affinity_type IN ('Ki', 'Kd', 'IC50', 'EC50')
    GROUP BY target_name, uniprot_raw
    ORDER BY median_p DESC, uniprot_raw
    
"""
await replay.call(3, 'run_sql', {'sql': sql},
                  sha256='9753fde3a8de8252c11ebdacfc250518da4d723e757b5c5465ecbf0315d44cdd')
````

````python
# Call 4 · run_sql · ran 2026-11-24 04:55:00 UTC
sql = """
    -- Imatinib's EGFR and ERBB2 measurements by relation and value
    SELECT uniprot_raw, affinity_type, relation, value, count(*) AS n
    FROM activity
    WHERE monomerid = 13530 AND uniprot_raw IN ('P00533', 'P04626')
    GROUP BY ALL
    ORDER BY uniprot_raw, affinity_type, relation, value
    
"""
await replay.call(4, 'run_sql', {'sql': sql},
                  sha256='f937fa6e020a4d16abaf771decf19caa9ac43a894dbf846298041ade901d0a75')
````

**🧠 Assistant**

**Imatinib** (Gleevec, STI-571) is in BindingDB as **BDBM13530**, with **1,156 measurements against 459 targets**. It is one of the most heavily profiled drugs in the database.

| Field | Value |
|---|---|
| BindingDB id | BDBM13530 |
| Other names | Gleevec, STI571, STI-571, CHEMBL941, PubChem cid_5291 |
| Formula / MW | C29H31N7O / 493.26 |
| InChIKey | KTUFNOKKBVMGRW-UHFFFAOYSA-N |
| SMILES | CN1CCN(Cc2ccc(cc2)C(=O)Nc2ccc(C)c(Nc3nccc(n3)-c3cccnc3)c2)CC1 |

The name search also returned two other records built on the same skeleton: **BDBM31341** (an "IMATINIB" entry with no measurements attached) and **BDBM50434581** (imatinib-CD3, a deuterated version with 2 measurements). A single drug can appear under several BindingDB ids, so check the measurement counts and use the id that carries the data (Example 1.2 shows a fourth record that the name search misses).

**Main targets.** These are the known clinical targets of imatinib. The table shows the **median** across all measurements for each human target, next to the single best value:

| Target | UniProt | Measurements | Median pAffinity | Best pAffinity |
|---|---|---|---|---|
| PDGFR-alpha | P16234 | 17 | 7.5 (~30 nM) | 8.7 |
| KIT | P10721 | 52 | 7.0 (~100 nM) | 7.9 |
| ABL1 | P00519 | 85 (7 censored) | 6.8 (~170 nM) | 9.0 |

DDR1 (Q08345), a known secondary target of imatinib, is also well supported: 9 measurements, median 7.4 (~40 nM), best 9.2.

(pAffinity = -log10 of the affinity in molar units, so 9 = 1 nM, 7 = 100 nM, 5 = 10 µM.)

**A lesson about "best" values.** The compound summary ranks targets by their single most potent measurement. On that ranking imatinib's top two targets are **ERBB2 (best pAffinity 10.2)** and **EGFR (9.96)**, which is misleading:

| Target | Measurements | Censored ("> value") | Median pAffinity | Best value |
|---|---|---|---|---|
| EGFR (P00533) | 25 | 21 (20 Kd > 10 µM, 1 IC50 > 100 µM) | 5.0 | one IC50 of 0.11 nM |
| ERBB2 (P04626) | 3 | 2 (Kd > 10 µM) | 5.0 | one IC50 of 0.06 nM |

For EGFR the three exact Kd values are all 7.6 µM; the sub-nanomolar IC50 is a single outlier against 24 measurements that say imatinib barely binds. When you judge how strongly a compound binds a target, look at the median and the number of measurements, not only the best value.

**How this was answered**

- `search_compounds` with the query "imatinib": finds compounds by name, synonym or identifier. Exact name matches rank first.
- `get_compound` with id 13530: returns identifiers, synonyms and a per-target summary of the most potent measurements.
- `run_sql`: computes the median affinity per target from the `activity` table, which holds one row per measurement, and lists the EGFR and ERBB2 values by relation:

```sql
SELECT target_name, uniprot_raw, count(*) AS n,
       sum(CASE WHEN relation = '>' THEN 1 ELSE 0 END) AS n_censored,
       round(median(p_affinity), 2) AS median_p,
       round(max(p_affinity), 2) AS best_p
FROM activity
WHERE monomerid = 13530
  AND uniprot_raw IN ('P00519', 'P10721', 'P16234', 'P04626', 'P00533', 'Q08345')
  AND affinity_type IN ('Ki', 'Kd', 'IC50', 'EC50')
GROUP BY target_name, uniprot_raw
ORDER BY median_p DESC, uniprot_raw
```
