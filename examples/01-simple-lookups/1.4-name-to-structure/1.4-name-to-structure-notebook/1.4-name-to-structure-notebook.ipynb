**🧠 Assistant**

All three names resolved to a structure, each through a **different source**:

| Name given | Resolved by | Compound | BindingDB record | Measurements | SMILES |
|---|---|---|---|---|---|
| Ibrance | BindingDB synonym | palbociclib | BDBM6309 | 154 | CC1=C(C(=O)N(c2c1cnc(n2)Nc3ccc(cn3)N4CCNCC4)C5CCCC5)C(=O)C |
| N-(3-chloro-4-fluorophenyl)-7-methoxy-6-(3-morpholin-4-ylpropoxy)quinazolin-4-amine | OPSIN (IUPAC name parser) | gefitinib | BDBM5447 | 1,177 | COc1cc2ncnc(Nc3ccc(F)c(Cl)c3)c2cc1OCCCN1CCOCC1 |
| 50-78-2 (CAS) | PubChem | aspirin | BDBM22360 | 137 | CC(=O)Oc1ccccc1C(=O)O |

**How the lookup works.** The tool tries three sources in order and stops at the first one that knows the name:

1. **BindingDB synonyms**: trade names, common names, ChEMBL ids and patent labels already in the database. "Ibrance" is a BindingDB synonym of palbociclib, and only one compound carries it, so the answer is not ambiguous.
2. **OPSIN**: a parser that turns systematic IUPAC names into structures. It runs offline and needs no database. BindingDB did not know the IUPAC string as written, but OPSIN built the structure from the name itself.
3. **PubChem**: name lookup on the web, which also handles CAS numbers. Neither BindingDB nor OPSIN knew 50-78-2; PubChem resolved it to aspirin (PubChem CID 2244).

When a structure comes from OPSIN or PubChem, the tool **matches it back to BindingDB by InChIKey**. For gefitinib it found three records with the same full InChIKey (XGALLCVXEZPNRQ-UHFFFAOYSA-N): BDBM5447 ("Iressa", 1,177 measurements), BDBM50338602 (also "Iressa", 6 measurements) and BDBM32367 ("GEFITINIB", no measurements). Use the record with the data. Aspirin has a single record, BDBM22360, with 137 measurements. Note that each record's preferred name is often a brand name ("Ibrance", "Iressa"), which shows why matching by structure is more reliable than matching by name.

**The lesson: resolve names to structures first.** Most of the powerful searches (similarity, substructure, exact structure) take a SMILES string, not a name. Converting a name to a SMILES string is usually the first step of a structure-based analysis. Watch for one pitfall: a generic label such as "compound 5" can belong to many BindingDB compounds from different papers. The tool then returns all candidates and sets `ambiguous=true`, and you should pick one by checking the structures.

- `chemical_name_to_smiles` three times, once per name, with the default source order (BindingDB, then OPSIN, then PubChem). Each reply lists the sources it tried and what each one returned.
