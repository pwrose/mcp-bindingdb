## Exchange 1

**👤 User**

I have a few identifiers for the same drug: CHEMBL941, PubChem CID 5291, BDBM13530 and the InChIKey block KTUFNOKKBVMGRW. Do they all find the same compound in BindingDB?

````python
# Call 1 · search_compounds · ran 2026-11-24 04:54:59 UTC
await replay.call(1, 'search_compounds', {'query': 'CHEMBL941'},
                  sha256='7963be8e713f17aaa52aa406c3c57b33c98259917aba3e74057995307440f1ec')
````

````python
# Call 2 · search_compounds · ran 2026-11-24 04:55:00 UTC
await replay.call(2, 'search_compounds', {'query': 'cid_5291'},
                  sha256='3224e9d305c02a073fa719edacbb003348dd26720c50c647d5e9dec44f7b5c05')
````

````python
# Call 3 · search_compounds · ran 2026-11-24 04:55:00 UTC
await replay.call(3, 'search_compounds', {'query': 'BDBM13530'},
                  sha256='32a3b365646fc9b78bc5ed433ddbd0462d6b6f5648274c064708abcf0a9ccd04')
````

````python
# Call 4 · search_compounds · ran 2026-11-24 04:55:00 UTC
await replay.call(4, 'search_compounds', {'query': 'KTUFNOKKBVMGRW'},
                  sha256='81dd9cd1ed2f5e55cacac3847cbe5b524c03c7a207f2d84f970f9381f0136f9b')
````

````python
# Call 5 · run_sql · ran 2026-11-24 04:55:00 UTC
sql = """
    -- Every BindingDB record on the imatinib skeleton (InChIKey first block KTUFNOKKBVMGRW), with its measurements and targets
    SELECT c.monomerid, 'BDBM' || c.monomerid AS bdbm_id, c.name, c.inchi_key, c.smiles,
           count(a.monomerid) AS n_measurements,
           count(DISTINCT coalesce(CAST(a.polymerid AS VARCHAR), 'c' || a.complexid)) AS n_targets
    FROM compound c LEFT JOIN activity a USING (monomerid)
    WHERE c.inchi_key LIKE 'KTUFNOKKBVMGRW-%'
    GROUP BY ALL
    ORDER BY n_measurements DESC, c.monomerid
    
"""
await replay.call(5, 'run_sql', {'sql': sql},
                  sha256='0408b2da57419f9f1a3425c706493659fc4a470c4588595c2744fd2d7cb72525')
````

**🧠 Assistant**

Yes. All four identifiers lead to **imatinib (BDBM13530)**, the record with 1,156 measurements against 459 targets. The InChIKey search also turns up **four more BindingDB records** built on the same skeleton, which the other identifiers miss.

| Query | Type | Top hit | Notes |
|---|---|---|---|
| CHEMBL941 | ChEMBL id | BDBM13530 | Exact match ranks first. The other 19 rows are substring matches (CHEMBL94191, CHEMBL94139, CHEMBL9416 ...) and should be ignored |
| cid_5291 | PubChem CID | BDBM13530 | BindingDB stores PubChem CIDs as `cid_<number>`; again, 19 substring matches (cid_5291664 ...) follow |
| BDBM13530 | BindingDB id | BDBM13530 | Only one hit |
| KTUFNOKKBVMGRW | InChIKey first block | 5 records | See below |

**What the InChIKey search adds.** An InChIKey has three blocks. The first 14 characters encode the molecular skeleton (atoms and connectivity). The second block encodes stereochemistry and isotopes, and the last character encodes protonation. Searching on the first block alone finds every record built on the same skeleton:

| BindingDB id | Name | Full InChIKey | Measurements | Targets | SMILES |
|---|---|---|---|---|---|
| BDBM13530 | Gleevec | KTUFNOKKBVMGRW-UHFFFAOYSA-N | 1,156 | 459 | CN1CCN(Cc2ccc(cc2)C(=O)Nc2ccc(C)c(Nc3nccc(n3)-c3cccnc3)c2)CC1 |
| BDBM86734 | Gleevec | KTUFNOKKBVMGRW-UHFFFAOYSA-N | 66 | 10 | Cc3ccc(NC(=O)c2ccc(CN1CCN(C)CC1)cc2)cc3Nc5nccc(c4cccnc4)n5 |
| BDBM50434581 | imatinib-CD3 | KTUFNOKKBVMGRW-BMSJAHLVSA-N | 2 | 2 | [2H]C([2H])([2H])N5CCN(Cc4ccc(C(=O)Nc3ccc(C)c(Nc2nccc(c1cccnc1)n2)c3)cc4)CC5 |
| BDBM50634633 | (no name) | KTUFNOKKBVMGRW-BMSJAHLVSA-N | 2 | 1 | [2H]C([2H])([2H])N5CCN(Cc4ccc(C(=O)Nc3ccc(C)c(Nc2nccc(c1cccnc1)n2)c3)cc4)CC5 |
| BDBM31341 | IMATINIB | KTUFNOKKBVMGRW-UHFFFAOYSA-N | 0 | 0 | Cc1ccc(cc1Nc2nccc(n2)c3cccnc3)NC(=O)c4ccc(cc4)CN5CCN(CC5)C |

**The lesson: one molecule can have several BindingDB records.** BindingDB creates records as data are deposited from different papers and patents. **BDBM86734** has the identical full InChIKey as BDBM13530 and carries 66 more imatinib measurements on 10 targets. A search on the name or ChEMBL id never shows it. The two deuterated records (second block `BMSJAHLVSA`) are isotope-labelled imatinib (a CD3 group on the piperazine), which is a different substance and usually should not be pooled.

If you want *all* data for a compound, search by the **full InChIKey** (same molecule, same isotopes and stereo) or by the **first block** (all stereoisomers, isotopologues and charge states), then decide which records to merge. Tools such as `find_targets_for_compound` accept an `inchi_key` argument for this.

**How this was answered**

- `search_compounds` four times, once per identifier. The same tool accepts names, ChEMBL ids, PubChem `cid_N`, BDBM ids and full or partial InChIKeys.
- `run_sql` to count measurements and targets for every record whose InChIKey starts with `KTUFNOKKBVMGRW-`:

```sql
SELECT c.monomerid, 'BDBM' || c.monomerid AS bdbm_id, c.name, c.inchi_key, c.smiles,
       count(a.monomerid) AS n_measurements,
       count(DISTINCT coalesce(CAST(a.polymerid AS VARCHAR), 'c' || a.complexid)) AS n_targets
FROM compound c LEFT JOIN activity a USING (monomerid)
WHERE c.inchi_key LIKE 'KTUFNOKKBVMGRW-%'
GROUP BY ALL
ORDER BY n_measurements DESC, c.monomerid
```
