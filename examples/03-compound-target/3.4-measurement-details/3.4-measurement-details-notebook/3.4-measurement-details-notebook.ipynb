**🧠 Assistant**

Gefitinib (Iressa, BDBM5447) has **256 measurements on wild-type EGFR** from 151 different papers and patents: 204 IC50, 42 Kd, 8 Ki and 2 EC50 values. They range from pAffinity 4.03 to 10.0, with a median of 7.85 (~14 nM). The most potent value, **IC50 = 0.1 nM**, is shared by four measurements from four papers (ki_result_id 50076446, 50087097, 51193991 and 51421148). This is the third of them in full:

| Curation route | Imported from ChEMBL (release 26) |

Gefitinib is the **reference compound** in that paper, not the subject. Its value comes from a FRET (Z'-Lyte) enzyme assay on an EGFR of "unknown origin". The first of the four (50076446, PMID 23792318, Eur J Med Chem 2013) is from the same group and the same kind of assay (Z'-Lyte, 1.5 h, Tyr 4 peptide substrate), so two of the four "independent" 0.1 nM values are not really independent. All four are ChEMBL imports with one-line assay descriptions.

For comparison, a fully annotated record (measurement 8648, IC50 = 1 nM). BindingDB curators extracted this one directly from the paper, so it carries the experimental conditions:

| Field | Value |
|---|---|
| Assay | "Kinase Inhibition Assay": radiometric [gamma-33P]ATP assay with peptide substrate, 10 min at 23 C, filter-plate scintillation counting, 11-point dose-response (0.0001-10 uM) |
| Conditions | pH 7.5, 296.15 K (23 C); enzyme 20 nM; substrate 50 uM ("EGFR Substrate Peptide") |
| Publication | Zhang YM et al., "Synthesis and SAR of potent EGFR/erbB2 dual inhibitors", Bioorg Med Chem Lett 2004, 14:111. PMID 14684309, doi:10.1016/j.bmcl.2003.10.010 |

**Lesson: every number has a provenance, so open it before you trust it.** BindingDB records come from two routes. Some are curated directly from articles and patents, and these often carry pH, temperature, enzyme and substrate concentrations and a detailed protocol. Others are imported from ChEMBL and usually carry a one-line assay description. Across the whole database, only about 212,000 of 3.25 million measurements (6.5%) report a pH. IC50 values depend on the assay format and on ATP and enzyme concentrations, which helps explain why gefitinib's EGFR values span six orders of magnitude (pAffinity 4.03-10.0). The "best" value of 0.1 nM is about 140-fold more potent than the median. Use `get_activity` to see where a surprising number came from, and cite the original paper, not just the database.
