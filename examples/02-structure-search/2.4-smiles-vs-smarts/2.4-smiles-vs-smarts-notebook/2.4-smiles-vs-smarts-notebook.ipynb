**Lesson: SMILES describes a molecule, SMARTS describes a pattern.** When you search with a SMILES string, the server matches it as a pattern made of the heavy atoms and bonds you wrote. Hydrogen counts are ignored, so N (or even [NH2]) matches a nitrogen with any number of substituents. To make a property part of the query, write SMARTS, which lets you state it per atom:

- [NX3;H2]: a three-connected nitrogen with exactly two H (a primary amine).
- [NX3;H1]: exactly one H (a secondary amine or amide NH).
- [n] vs [N]: aromatic vs non-aromatic nitrogen.
- [#7]: any nitrogen, whatever its form.

With query_type="auto" the server falls back to SMARTS when a query is not valid SMILES. If you need H counts enforced, set query_type="smarts" yourself.
