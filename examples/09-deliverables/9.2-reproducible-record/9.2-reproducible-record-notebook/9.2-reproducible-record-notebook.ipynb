Keep the notebook in its folder, which holds everything it needs: `requirements.txt` (mcp-bindingdb at its commit and the versions of the packages it needs) and the recorded results in `data/`. With [uv](https://docs.astral.sh/uv/), from that folder (replace the path with yours):

```bash

uv run --no-project --with-requirements requirements.txt --with jupyterlab jupyter lab 9.2-reproducible-record-notebook.ipynb
```

**🧠 Assistant**

All three drugs prefer CDK4 and CDK6, but **abemaciclib is clearly the least selective** within the CDK family:

| Drug | CDK4 | CDK6 | CDK9 | CDK1 | CDK2 | CDK4 vs CDK2 |
|---|---|---|---|---|---|---|
| Palbociclib | 11 nM (n=46) | 15 nM (n=34) | 841 nM (n=6) | inactive (1 exact, 8 "> value") | 8.9 µM (n=14) | ~800× |
| Ribociclib | 10 nM (n=22) | 39 nM (n=16) | 195 nM (n=4) | inactive (1 exact, 5 "> value") | 76 µM (n=2) | ~7,600× |
| Abemaciclib | 2 nM (n=39) | 9 nM (n=38) | 58 nM (n=12) | 372 nM (n=17) | 316 nM (n=21) | ~160× |

(Medians of exact Ki, Kd and IC50 values; "inactive" means fewer than 2 exact values, with most measurements reported only as "> value". Abemaciclib was also measured on CDK5, 288 nM (n=11), and CDK7, 2.1 µM (n=9); palbociclib and ribociclib have only "> value" results on CDK5 and CDK7.)

Abemaciclib binds CDK9 only about 30-fold more weakly than CDK4, and CDK1 and CDK2 about 160- to 190-fold more weakly. Palbociclib and ribociclib keep a window of roughly three orders of magnitude over CDK2 and CDK9 (palbociclib ~75× over CDK9), and are essentially inactive on CDK1, CDK5 and CDK7. Ribociclib's CDK2 value rests on just two exact measurements (7 more are "> value"), so its 7,600× window is the least certain number in the table. Because "> value" results are left out of the medians, the medians for the weak targets are if anything too potent: palbociclib also has 10 censored results on CDK2 besides its 14 exact ones.

**The lesson: a reproducible record shows the choices, not just the answer.** The tool calls below contain the exact SQL, so a reader can see every decision behind the table:

- **Which records count as "CDK1", "CDK2", ...** BindingDB names CDK targets in many ways: "Cyclin-dependent kinase 6/G1/S-specific cyclin-D3", "CDK6/cyclin D1", "Cyclin-A2/Cyclin-dependent kinase 2", and even "Cyclin-dependent kinase/G2/mitotic-specific cyclin- 1", which is the CDK1/cyclin B1 complex. A pattern on the target name misses that last record, which holds 9 of abemaciclib's 17 exact CDK1 values, and would count the mixed record "Cyclin-dependent kinase 4/6" as CDK4. So the query takes the CDK from the target's protein chains instead: a chain counts as CDK*n* when one of its BindingDB names is exactly "CDK*n*", and records with two different CDKs are skipped. The second query lists every record assigned to each CDK, with exact and censored counts per drug, so the assignment can be checked line by line.
- **Pooling.** Plain kinases and CDK/cyclin complexes are pooled, and mutant records (names with "[...]") are excluded.
- **Which values.** Only exact ('=') Ki, Kd and IC50 values are used; EC50 values (often cell-based) are left out. Censored values such as "> 10 µM" are excluded (see Example 8.4).

A colleague who disagrees with any of these choices can change one line of SQL and rerun it. A results table on its own does not allow that. The choices matter here: the CDK1 column for abemaciclib (372 nM from 17 values) would read 56 nM from 9 values with a name pattern that misses the CDK1/cyclin B1 record.

**How this was answered**

- `search_compounds` for ribociclib, abemaciclib and palbociclib to get their BindingDB ids (BDBM148264, BDBM50110183 and BDBM6309).
- `run_sql`: listed the kinase target names measured for the three drugs, to see how BindingDB names the CDK records.
- `run_sql`: median pAffinity per drug and CDK, as described above.
- `run_sql`: the target records counted as each CDK, with exact and censored counts per drug.
- `run_sql`: the same medians with a name pattern on the target name instead (CDK number taken from "cyclin-dependent kinase *n*" or "CDK*n*", all exact values), for comparison: it gives abemaciclib 56 nM on CDK1 (n=9) and pools 43 values as CDK4.
- `create_transcript` with `include_tool_calls=true`, listing only the calls that belong to this analysis. The server logs every BindingDB call it runs, with its arguments and full result.

**How to make your own reproducible record.** Ask for it in plain words: "include the tool calls", "with the queries and results", or "a reproducible record". It is cleanest to start the analysis with a fresh log ("reset the tool log first"). Otherwise the assistant picks your calls out of the shared log by index. Exploratory calls (here, two lookups of the components of the oddly named complexes) are left out, so the record shows the path to the answer, not every detour.
